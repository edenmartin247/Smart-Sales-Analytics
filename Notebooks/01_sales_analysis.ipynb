In this project, EDA is performed on the SuperStore Orders dataset containing 51,290 records and 25 columns.

- Analyze overall sales and profit performance.

- Examine discount and shipping performance.

### 2. Overall Business KPI Analysis

Key Performance Indicators (KPIs) are measurable values used to evaluate the overall performance of a business. In this section, we calculate total sales, total profit, total orders, average order value, and profit margin to understand the business's overall performance.

```python
total_sales = df["sales"].sum()

total_profit = df["profit"].sum()

total_orders = df["order_id"].nunique()

order_sales = df.groupby("order_id")["sales"].sum()

average_order_value = order_sales.mean()

profit_margin = (total_profit / total_sales) * 100

print("===== SALES PERFORMANCE =====")

print("Total Sales:", round(total_sales, 2))

print("Total Profit:", round(total_profit, 2))

print("Total Orders:", total_orders)

print("Average Order Value:", round(average_order_value, 2))

print("Profit Margin:", round(profit_margin, 2), "%")
```

Key Business Insights
- The dataset contains 25,035 unique orders.
- Total sales reached 12.64 million.
- Total profit was approximately 1.47 million.
- Average order value was 505.01.
- Overall profit margin was 11.62%.
- The business generated positive overall profit.

### 3. Category and Sub-category Analysis

This analysis examines sales and profit across different product categories and sub-categories. It helps identify the best-performing product groups and understand their contribution to overall business performance.

```python
category_analysis = df.groupby("category").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    total_quantity=("quantity", "sum")
).sort_values(by="total_sales", ascending=False)

print("===== CATEGORY PERFORMANCE =====")

print(category_analysis.round(2))
```

Category Analysis – Key Insights
- Technology is the leading category in terms of total sales and profit.
- Furniture generates the lowest profit among the three categories.
- Office Supplies has the highest total quantity sold.
- High sales volume does not necessarily mean the highest profit.
- Category-level analysis helps identify profitable product groups.

### 3.2 Sub-category Performance Analysis

Sub-category analysis provides a more detailed view of product performance. It helps identify which specific product groups contribute the most to sales and profit.

```python
subcategory_analysis = df.groupby("sub_category").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    total_quantity=("quantity", "sum")
).sort_values(by="total_sales", ascending=False)

print("===== SUB-CATEGORY PERFORMANCE =====")

print(subcategory_analysis.round(2))
```

Sub-category Analysis – Key Insights
- Phones are the highest revenue-generating sub-category.
- Copiers generate the highest total profit.
- Binders have the highest sales quantity.
- Tables generate a negative total profit, indicating a potential loss-making product group.
- Labels have the lowest total sales.
- Product-level analysis helps identify profitable and loss-making sub-categories.

# 4. Regional and Market Analysis
Regional and Market Analysis helps understand how sales and profit vary across different geographical locations. It identifies high-performing regions and markets and helps businesses understand their geographical performance.

### 9.1 Shipping Mode-wise Performance

Key business insights

3. First Class takes an average of 2.18 days.
4. Second Class takes an average of 3.23 days.
5. Standard Class has the highest total shipping cost, but it also handles the most orders.

### 9.2 Shipping Mode-wise Profit Analysis

```python
plt.figure(figsize=(10, 5))

shipping_performance["total_profit"].sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Shipping Mode-wise Total Profit")
plt.xlabel("Shipping Mode")
plt.ylabel("Total Profit")
plt.xticks(rotation=20)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
